### Calculating Experience_year1 column

```python
#year = df_name["recruitment_date1"].dt.year   
this_year = jdatetime.date.today().year       #age are ye doonaro estefade kon
df_Kourosh["Experienced_year1"] = this_year - year
```

```python
mean_experience = round(df_Kourosh["Experienced_year1"].mean(),2)
```

```python
fig = px.histogram(
    df_Kourosh, x = "Experienced_year1", nbins = 12, title = "Distribution of Employee Experience",
    labels = {
    "Experienced_year1":"Work Experience (year)", "count":"Number of Employees"
    },
color_discrete_sequence=["steelblue"
                        ]
)

# khate amoodi ghermez monaaaa
fig.add_vline(
    x = mean_experience,line_dash = "dash",line_color = "red",line_width = 5,annotation_text = f"Mean={mean_experience:.2f} years", 
    annotation_position = "top", annotation_font_color = "red" , annotation_font_size = 15
)

fig.update_layout(
    template = "plotly_white",xaxis_title="work Experience (years)", yaxis_title="Number of Employees",
    title_x = 0.2,font = dict(size = 11),bargap = 0.5
)
fig.show()
```
